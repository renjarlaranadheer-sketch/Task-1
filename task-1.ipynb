**#  Missing values**

```python
print("Missing before:", df.isnull().sum())

# Fill text with "unknown"
text_cols = df.select_dtypes(include='object').columns
df[text_cols] = df[text_cols].fillna("unknown")

# Fill numbers with median
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
df=df.drop_duplicates()
print(df)
```

**Standardize text (lower + strip)**

```python
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
print(df[col])
```

**Convert any column with "date" in its name to datetime**

```python
for col in df.columns:
    if "date" in col:
        df[col] = pd.to_datetime(df[col], errors='coerce')
print(df[col])
```

**Save cleaned file**

```python
df.to_csv("/kaggle/working/cleaned.csv", index=False)

print("Cleaning complete. Saved as cleaned.csv")
```
